```python
df_clean = df[[
    "Geographic area",
    "Indicator",
    "Sex",
    "Wealth Quintile",
    "DATA_SOURCE",
    "COUNTRY_NOTES",
    "Unit of measure",
    "TIME_PERIOD",
    "OBS_VALUE",
    "LOWER_BOUND",
    "UPPER_BOUND",
    "Observation Status"
]].copy()
```

```python
df_clean = df_clean.rename(columns={
    "Geographic area": "country",
    "Indicator": "indicator",
    "Sex": "sex",
    "Wealth Quintile": "wealth quintile",
    "DATA_SOURCE": "data source",
    "COUNTRY_NOTES": "country notes",
    "Unit of measure": "unit measure",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "obs value",
    "LOWER_BOUND": "lower bound",
    "UPPER_BOUND": "upper bound",
    "Observation Status": "observation status"
})
```

```python
df_clean = df_clean.dropna(subset=["country", "indicator", "year", "obs value"])
```

```python
text_cols = ["country", "indicator", "sex", "wealth quintile", "data source", "unit measure", "observation status"]

for col in text_cols:
    df_clean[col] = df_clean[col].apply(lambda x: x.strip() if isinstance(x, str) else x)
```

```python
numeric_cols = ["year", "obs value", "lower bound", "upper bound"]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
```

```python
df_clean["year"] = df_clean["year"].astype(int)
```
